```python
import numpy as np
import requests
from io import BytesIO
from PIL import Image
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing.image import img_to_array
```

```python
# --- CONFIGURAZIONE ---
# Utilizzare l'architettura VGG16 al posto di MobileNetV2.
# Cambiare l'immagine di input (scegliete un soggetto diverso da una fragola) (esempio: un elefante da Wikimedia Commons)
IMMAGINE_URL = "https://upload.wikimedia.org/wikipedia/commons/3/37/African_Bush_Elephant.jpg"

print(f"URL impostato: {IMMAGINE_URL}")
```

```python
# --- DEFINIZIONE DELLE FUNZIONI ---
# Carichiamo VGG16 addestrata su ImageNet. 
# include_top=False: Escludiamo i layer Fully Connected finali (il classificatore).
def carica_modello():
    modello = VGG16(weights='imagenet', include_top=True)
    return modello

def carica_immagine(url, target_size=(224, 224)):
    response = requests.get(url)    # 
    img = Image.open(BytesIO(response.content)).convert('RGB')
    img = img.resize(target_size)  # resize invariato: VGG16 usa anch'essa 224x224
    array = img_to_array(img)
    array = np.expand_dims(array, axis=0)
    array = preprocess_input(array)  # <-- preprocessing specifico VGG16 (sottrazione media RGB ImageNet)
    return array
```
